```python
orders = orders.drop(
    columns=['campaign_name', 'coupon_code']
)
```

```python
orders['return_status'] = orders['return_status'].fillna(
    'Not Returned'
)

orders['return_reason'] = orders['return_reason'].fillna(
    'Not Applicable'
)
```

```python
orders['review_sentiment'] = orders['review_sentiment'].fillna(
    'No Review'
)

orders['customer_review'] = orders['customer_review'].fillna(
    'No Review'
)
```

```python
categorical_columns = [
    'order_status',
    'sales_channel',
    'gender',
    'customer_segment',
    'customer_type',
    'customer_state',
    'region',
    'payment_method',
    'payment_status',
    'shipping_method',
    'warehouse',
    'delivery_status',
    'return_status',
    'return_reason',
    'review_sentiment',
    'marketing_channel',
    'is_repeat_customer'
]

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(orders[col].value_counts(dropna=False))
```

```python
# Convert order_date to datetime
orders['order_date'] = pd.to_datetime(
    orders['order_date'],
    errors='coerce'
)
```

```python
print("Minimum order date:", orders['order_date'].min())
print("Maximum order date:", orders['order_date'].max())

print(
    "Invalid order dates:",
    orders['order_date'].isna().sum()
)
```

```python
print(
    "Invalid order dates:",
    orders['order_date'].isna().sum()
)
```

```python
orders['order_time'] = pd.to_datetime(
    orders['order_time'],
    format='%H:%M:%S',
    errors='coerce'
).dt.time
```

```python
orders['order_time'].head(10)
```

```python
orders['order_year'] = orders['order_date'].dt.year
orders['order_month'] = orders['order_date'].dt.month
orders['order_month_name'] = orders['order_date'].dt.month_name()
orders['order_quarter'] = orders['order_date'].dt.quarter
orders['order_day'] = orders['order_date'].dt.day
orders['order_day_name'] = orders['order_date'].dt.day_name()
```

```python
orders[
    [
        'gross_sales',
        'discount_amount',
        'tax_amount',
        'shipping_cost',
        'net_sales',
        'product_cost',
        'profit'
    ]
].describe()
```

```python
orders['delivery_variance'] = np.where(
    orders['delivery_days'].notna() &
    orders['estimated_delivery_days'].notna(),
    orders['delivery_days'] -
    orders['estimated_delivery_days'],
    np.nan
)
```

```python
orders['delivery_performance'] = np.select(
    [
        orders['delivery_variance'] < 0,
        orders['delivery_variance'] == 0,
        orders['delivery_variance'] > 0
    ],
    [
        'Early',
        'On Time',
        'Late'
    ],
    default='Not Available'
)
```

```python
# Rebuild datasets after cleaning
datasets = {
    "orders": orders,
    "order_items": order_items,
    "product_catalog": product_catalog,
    "customer_master": customer_master
}

# Final data-quality overview
for name, df in datasets.items():
    print(f"\n{'='*60}")
    print(name.upper())
    print(f"{'='*60}")
    
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Duplicate rows:", df.duplicated().sum())
    
    missing = df.isnull().sum()
    missing = missing[missing > 0].sort_values(ascending=False)
    
    print("\nMissing values:")
    
    if missing.empty:
        print("No missing values")
    else:
        print(missing)
```

```python
# Orders → Customer Master

order_customer_ids = set(
    orders['customer_id'].dropna()
)

master_customer_ids = set(
    customer_master['customer_id'].dropna()
)

missing_customers = order_customer_ids - master_customer_ids

print(
    "Customer IDs missing from customer_master:",
    len(missing_customers)
)
```

```python
# Order Items → Product Catalog

item_product_ids = set(
    order_items['product_id'].dropna()
)

catalog_product_ids = set(
    product_catalog['product_id'].dropna()
)

missing_products = item_product_ids - catalog_product_ids

print(
    "Product IDs missing from product_catalog:",
    len(missing_products)
)
```

```python
financial_columns = [
    'gross_sales',
    'discount_amount',
    'tax_amount',
    'shipping_cost',
    'net_sales',
    'product_cost',
    'profit',
    'profit_margin_percentage'
]

print("Financial Summary:")
display(orders[financial_columns].describe().T)
```

```python
print("\nNegative values:")

for col in financial_columns:
    print(
        f"{col}:",
        (orders[col] < 0).sum()
    )
```

```python
print("\nProfit Margin Summary:")
print(orders['profit_margin_percentage'].describe())
```

```python
# Validate profit margin calculation

orders['calculated_profit_margin'] = (
    orders['profit'] / orders['net_sales']
) * 100

orders['margin_difference'] = (
    orders['profit_margin_percentage']
    - orders['calculated_profit_margin']
)

print(
    "Maximum margin difference:",
    orders['margin_difference'].abs().max()
)

print(
    "Average margin difference:",
    orders['margin_difference'].abs().mean()
)
```

```python
print(
    orders[
        [
            'profit',
            'net_sales',
            'profit_margin_percentage',
            'calculated_profit_margin',
            'margin_difference'
        ]
    ].head(10)
)
```

```python
# Order Items → Orders

order_ids_orders = set(
    orders['order_id'].dropna()
)

order_ids_items = set(
    order_items['order_id'].dropna()
)

missing_orders = order_ids_items - order_ids_orders

print(
    "Order IDs in order_items but missing from orders:",
    len(missing_orders)
)
```
